Problem Statement 1:Smart Parking Lot Management System

```python
def parking_lot_management_system(capacity,logs):

  #count IN and OUT
  in_count = 0
  out_count = 0
  for entry in logs:
    if entry == "IN":
      in_count += 1
    elif entry == "OUT":
      out_count +=1

  #Calculate Currently parked vehicles
  total_vehicles = in_count - out_count
  print("Currently Parked Vehciles",total_vehicles)

  #Identify peak parking usage
  #Alert if parking exceeds capacity
  if total_vehicles > capacity:
    print("Parking Status:Full")
  else:
    print("Parking Status:Available")

#Input
capacity=50
logs=["IN","IN","IN","OUT","IN","IN","OUT"]
parking_lot_management_system(capacity,logs)
```


 Problem Statement 2:Online Food Delivery Time Estimator

```python
def food_delivery_time_estimator(distance,traffic,weather):

  #Distance(km)
  time=distance*5

  #add delay for traffic
  if traffic == "HIGH":
    time+=15
  elif traffic == "MODERATE":
    time+=10

  #add delay for weather
  if weather == "RAIN":
    time+=10

  print("Estimated Delivery Time:",time,"minutes")

food_delivery_time_estimator(8,"High","Rainy")
```


 Problem Statement 3:Movie Theatre Seat Occupancy Analyzer


```python
def theatre_occupancy(total_seats,booked_seats):

  #calculate occupancy percentage
  occupancy=(len(booked_seats)/total_seats)*100
  print("Occupancy:",occupancy,"%")

  #Determines if show is Housefull
  if occupancy == 100:
    print("Show status:Housefull")
  elif occupancy >= 75:
    print("Show status:Almost Full")
  else:
    print("Show status:Available")

#Suggests opening additional shows
total_seats=200
Booked_seats=[1]*150
theatre_occupancy(total_seats,Booked_seats)
```

Problem Statement 4:Cloud Server Load Classification System


```python
def server_load(cpu_readings):
  total=0
  for value in cpu_readings:
    total+=value

  #Average CPU < 50% → Normal
  average=total/len(cpu_readings)
  print("Average CPU Load:",average,"%")

  #50%–80% → Warning
  if average < 50:
    print("Server Status:Normal")
  elif average >= 50 and average < 80:
    print("Server Status:Warning")

  #80% → Critical
  else:
    print("Server Status:Critical")

cpu= [45, 60, 70, 85, 90]
server_load(cpu)
```

PROBLEM 5: Smart Classroom Resource Usage Monitor

```python
def classroom_monitor(resources):

  #overusage
  overused=[]
  for name, value in resources.items():
    if value > 8:
      overused.append(name)
  #Alert of excess consumtions
  if overused:
    print("Overused Resources:",",".join(overused))
    print("Energy Alert:Yes")
  else:
    print("Energy Alert:No")
#resources
usage={
    "Projector":6,
    "AC":9,
    "Lights":4
}

classroom_monitor(usage)
```

 Problem Statement 6: Online Event Registration Capacity Controller


```python
def event_registration(capacity,registrations):

#Tracking registrations
  if registrations > capacity:

#Preventing overbooking
    confirmed = capacity
    waitlisted = registrations - capacity
    status ="Closed"

#Triggering waitlist mode
  else:
    confirmed = registrations
    waitlisted = 0
    status = "Open"

  print("Confirmed Registration:",confirmed)
  print("Waitlisted Users:",waitlisted)
  print("Registration Status:",status)
# Note: `capacity` and `registrations` are not defined in the global scope here.
# For testing, you might add example values like:
capacity = 100
registrations = 105
event_registration(capacity,registrations)
```
