Problem 1: Mobile Recharge Validation System

```python
def mobile_recharge_validation():
    mobile_number = input("Enter your mobile number: ")
    valid_plans=[199,299,399,599]
    #Use function
    while True:
      amount = int(input("Enter the recharge amount: "))
      #Use for loop on dictionary
      if amount < 50:
        print("Invalid recharge amount")
        continue
      elif amount in valid_plans:
        print("You have successfully recharged Rs.",amount)
        break
      else:
        print("Invalid recharge amount")
        continue
mobile_recharge_validation()
```

Problem 2: Inventory Reorder Alert System


```python
def inventory_reorder_alert():
#Use function
  product = {
    "salt":18,
    "Rice": 10,
    "oil": 5,
    "sugar":20
  }
#Use a for loop on a dictionary
  for item in product:
    if product[item]<15:
      print(item,"-> Reorder Alert")
    else:
      print(item,"->Stock Ok")
inventory_reorder_alert()
```

 Problem 3: Student Result Processing System

```python
def check_result(marks):
  total=sum(marks)
  average=total/len(marks)
  #Use loop to calculate total
  print("Total Marks:",total)
  print("Average Marks:",average)
  if average>=50:
    result_str = "Pass"
  else:
    result_str = "Fail"
  print("Result:", result_str)
  #Return result from function
  return result_str

marks=[60,70,40,55]
result=check_result(marks)
print(result)
```

Problem 4: Cab Fare Estimator with Retry Option

```python
def calculate_fare(distance,peak):
    #Function for fare calculation
    fare=50+(12*distance)
    if peak==True:
        fare=fare+(fare*0.25)
    return fare
#While loop for retry
while True:
  km=float(input("Enter the distance in km: "))
  peak_hour=input("Is it a peak hour? (yes/no): ")
  total_fare=calculate_fare(km,peak_hour=="yes")
  print("Total fare is Rs.",total_fare)
  retry=input("Do you want to calculate another fare? (yes/no): ")
  if retry=="no":
    break
```

Problem 5: Employee Attendance Eligibility Checker

```python
def check_attendance(attendance):
  present=0
#Use loop to count present days
  for day in attendance:
    if day=="p":
      present+=1
  total_days=len(attendance)
  percentage=(present/total_days)*100
  print("Attendance percentage:",percentage)
#Use return statement
  if percentage>=75:
    return "Eligible"
  else:
    return "Not Eligible"
attendance=["p","a","p","a","p","p","a"]
result=check_attendance(attendance)
print(result)
```

Problem 6: Password Strength Checker




```python
def check_password_strength(password):
  has_digit=False
  has_uppercase=False
  has_lowercase=False
  has_special=False
  special_characters="@#$"
  # Use loop over string and Use conditions inside function
  if len(password)<8:
    return "Weak password"

  for char in password:
    if char.isdigit():
      has_digit=True
    elif char.isupper():
      has_uppercase=True
    elif char.islower():
      has_lowercase=True
    elif char in special_characters:
      has_special=True

  # Check all conditions after iterating through the password
  if has_digit and has_uppercase and has_lowercase and has_special:
    return "Strong password"
  else:
    return "Weak password"

password=input("Enter a password:")
result=check_password_strength(password)
print(result)
```
