Problem 1:Unique words in a sentence

```python
#input
sentence= ("Python is easy and Python is powerful")

#split the sentence into individual words
words=sentence.split()

#convert the list of words into a set to remove duplicates
unique_words=set(words)

#print both the set and its length
print("Unique words count:",len(unique_words))
print("Unique words:",unique_words)

```

Problem 2: Highest Salary from Employee Data


```python
#Employees data
employees = {
    "Ravi":75000,
    "Anita":68000,
    "Kiran":72000
}
#Compare salary values to find the highest
highest_salary=max(employees.values())

#find the employee name
highest_salary_employee=[name for name,salary in employees.items() if salary==highest_salary]
print("Highest Salary:",highest_salary_employee[0],"-",highest_salary)
```

Problem 3:  Find Maximum and Minimum Values


```python
#list of numbers
numbers=[45,22,89,10,66]

#Initialize both maximum and minimum with the first element of the list.
maximum=minimum=numbers[0]

#use a loop to compare each element
for number in numbers:
  if number>maximum:
    maximum=number
  elif number<minimum:
    minimum=number

#print list,max,min values
print("List:",numbers)
print("Maximum:",maximum)
print("Minimum:",minimum)

```

Problem 4: Count Products Above a Price Threshold

```python
#Prices
prices=[450,1200,899,1500,300]

#Initialize a counter for products above 1000
count_above_1000 = 0

#use a loop to compare each price with 1000
for price in prices:
  if price > 1000:
    count_above_1000 += 1

#now print products above 1000
print("Products above 1000:", count_above_1000)
```

Problem 5: Calculate Attendance Percentage

```python
#Attendance
attendance= ["P", "P", "A", "P", "P"]

#count the number of "P" values
total_days=len(attendance)
present_days=attendance.count("P")

#divide by the total number of days
attendance_percentage=(present_days/total_days)*100

#now print attendance percentage
print("Attendance Percentage:",attendance_percentage)
```

Problem 7: Count Character Frequency

```python
#given string
string="python"

#use a dictionary
char_count={}

#Increment the count for each character occurrence
for char in string:
  if char in char_count:
    char_count[char]+=1
  else:
    char_count[char]=1
print(char_count)
```

Problem 10: Calculate Average Marks


```python
#Marks
student_marks=[45, 22, 89, 10, 66]

#calculate sum of all the values
total_marks=sum(student_marks)

#dividing by the total number of marks
average_marks=total_marks/len(student_marks)

#now print average marks
print("Average Marks:",average_marks)
```
