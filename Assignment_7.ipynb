Problem 1: Smart Light Controller

```python
#creating class
class SmartLightController:
    def __init__(self,name):
        self.name=name
        self.status="OFF"

    #Method to turn ON
    def turn_on(self):
        self.status="ON"
        print(f"{self.name} is {self.status}")

    #Methods to turn OFF
    def turn_off(self):
        self.status="OFF"
        print(f"{self.name} is {self.status}")

light=SmartLightController("Bedroom Light")
light.turn_on()
#light.turn_off()
```

Problem 2: Employee ID Card System


```python
#creating class for employee
class EmployeeIDCard:
    def __init__(self,name,emp_id,department):
        self.name=name
        self.emp_id=emp_id
        self.department=department
#Details that needs to be displayed in the ID card
    def display_card(self):
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.department}")
emp=EmployeeIDCard("Rahul","EMP102","IT")
emp.display_card()
```

Problem 3: Mobile Contact Record


```python
#creating class
class MobileContactRecord:
    def __init__(self,name,phone_number):
        self.name=name
        self.phone_number=phone_number

#contact details need to be displayed
    def display_contact(self):
        print("Contact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone_number}")

contact=MobileContactRecord("Anita","9876543210")
contact.display_contact()
```

Problem 4: Product Price Tag Generator

```python
#creating class
class ProductPriceTag:
    def __init__(self,name,price):
        self.name=name
        self.price=price

#Price tag
    def generate_tag(self):
        print("Product:",self.name)
        print("Price:₹",self.price)
product=ProductPriceTag("Headphones",2499)
product.generate_tag()
```

Problem 5: Movie Rating Display System

```python
#creating class
class MovieRatingDisplay:
    def __init__(self,name,rating):
        self.name=name
        self.rating=rating

#Rating display
    def display_rating(self):
        print("Movie:",self.name)
        print("Rating:",self.rating,"/5")
movie=MovieRatingDisplay("Inception",4.8)
movie.display_rating()
```

Problem 6: Delivery Address Manager

```python
#creating class
class DeliveryAddressManager:
    def __init__(self,name,address):
        self.name=name
        self.address=address

    #delivery details
    def display_address(self):
        print("Delivery Details")
        print("Customer:",self.name)
        print("Address:",self.address)
delivery=DeliveryAddressManager("Suman","Hyderabad")
delivery.display_address()
```
